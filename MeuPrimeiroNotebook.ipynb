# **Exercício**

Crie um código Python para receber os valores (reais) dos dois catetos (b e c) e retornar a hipotenusa (a)

```python
b = float(input('Digite o primeiro cateto: '))
c = float(input('Digite o segundo cateto: '))

a = (b**2 + c**2)**0.5

print(f'O valor da hipotenusa é: {a: .2f}')
```

```python
j = int(input('Valor de j: '))
k = int(input('Valor de k: '))

if j > k:
  # indentação obrigatória
  print(f'{j} é maior do que {k}')

if j < k:
  print(f'{j} é menor do que {k}')

if j == k:
  print(f'{j} é igual a {k}')

# != ---> diferente de
```
